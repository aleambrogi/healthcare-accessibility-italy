```python
import os
from pathlib import Path

# Get the current notebook's directory
notebook_dir = Path().absolute()

# Get the parent directory of the current notebook's folder
parent_dir = notebook_dir.parent

# Change the working directory to the parent folder
os.chdir(parent_dir)

# Print to verify the change
print("New Working Directory:", os.getcwd())
```

```python
import geopandas as gpd
import pandas as pd

# Percorso alla cartella con gli 11 shapefile
shapefile_folder = os.getcwd()

# Trova tutti i file .shp dentro la cartella (anche se in sottocartelle)
shapefile_paths = []
for root, dirs, files in os.walk(shapefile_folder):
    for file in files:
        if file.endswith(".shp"):
            shapefile_paths.append(os.path.join(root, file))

# Carica e unisci tutti gli shapefile
lista_gdf = [gpd.read_file(path) for path in shapefile_paths]
gdf_zone = gpd.GeoDataFrame(pd.concat(lista_gdf, ignore_index=True), crs=lista_gdf[0].crs)

print(f"✅ Caricati {len(lista_gdf)} shapefile. Totale geometrie: {len(gdf_zone)}")
```
